```python
# 1. High Level API의 핵심 모듈인 pipeline을 불러온다.
from transformers import pipeline
```

```python
model_id = "Bllossom/llama-3.2-Korean-Bllossom-3B"
```

```python
# 2. pipeline을 설정한다.
pipe = pipeline(
    task = "text-generation",
    model = model_id,
    device_map = "cpu",
    dtype = "auto",
    do_sample = True,
    top_p = 0.9,
    temperature = 0.7
)
```

```python
# 3. pipeline으로 질문에 대한 답변을 생성한다.
text = "철수가 10개의 사과를 가지고 있었는데 영희가 3개를 가져가면 철수에게 남은 사과의 갯수는 몇개인가요?"
result = pipe(text, max_new_tokens=100)
print(result[0]["generated_text"])
```
