```python
# 1. 필용한 라이브러리를 불러온다.
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM
```

```python
# 2. 모델명을 설정하고 토크나이저와 모델을 불러온다.
model_id = "Bllossom/llama-3.2-Korean-Bllossom-3B"
tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForCausalLM.from_pretrained(model_id)
```

```python
tokenizer.pad_token = tokenizer.eos_token
model.config.pad_token_id = tokenizer.eos_token_id
```

```python
question = "철수가 10개의 사과를 가지고 있었는데 영희가 3개를 가져가면 철수에게 남은 사과의 갯수는 몇개인가요?"
inputs = tokenizer(text=question, padding=True, return_tensors="pt")
```

```python
print(inputs)
```

```python
print(inputs.input_ids)
```

```python
print(inputs.attention_mask)
```

```python
# 5. 전처리 작업을 마친 질문을 모델에 넣어 답변 텍스트를 생성한다.
model.eval()
with torch.no_grad():
    output_ids = model.generate(
        input_ids = inputs["input_ids"],
        attention_mask = inputs["attention_mask"],        
        max_new_tokens = 100,
        do_sample = True,
        top_p = 0.9,
        temperature = 0.7
    )
```

```python
# 6. 결과를 확인할 때는 텐서 구조에서 0번째 요소를 추출하고, 사람이 읽을 수 있도록 디코딩을 한다.
result = tokenizer.decode(token_ids=output_ids[0], skip_special_tokens=True)
print(result)
```
