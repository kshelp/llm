```python
from langchain_ollama import ChatOllama 
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.output_parsers import StrOutputParser
from langchain_core.chat_history import InMemoryChatMessageHistory
from langchain_core.runnables import RunnableWithMessageHistory
```

```python
model_id = "timHan/llama3.2korean3B4QKM"
llm = ChatOllama(  
    model = model_id,
    base_url="http://localhost:11434",
    num_predict = 100,
    top_p = 0.9,
    temperature = 0.7
)
```

```python
prompt = ChatPromptTemplate.from_messages([
    ("system", "너는 친절한 한국어 대화 도우미야. 사용자의 말을 잘 기억하고 자연스럽게 답변해줘."),
    MessagesPlaceholder(variable_name="history"),
    ("human", "{user_input}")
])

parser = StrOutputParser()
base_chain = prompt | llm | parser
```

```python
session_histories = {}
def get_session_history(session_id):
    if session_id not in session_histories:
        session_histories[session_id] = InMemoryChatMessageHistory()
    return session_histories[session_id]
```

```python
chat_chain = RunnableWithMessageHistory(
    base_chain,
    get_session_history,               
    input_messages_key="user_input",   
    history_messages_key="history" 
)
```

```python
def chat(session_id, msg):
    response = chat_chain.invoke(
        {"user_input": msg},
        config={"configurable": {"session_id": session_id}}
    )
    return response
```

```python
print(chat("123", "나는 음식 중에 돈까스를 좋아한다는 사실을 기억해줘."))
```

```python
print(chat("123", "내가 좋아하는 음식이 뭐니?"))
```
